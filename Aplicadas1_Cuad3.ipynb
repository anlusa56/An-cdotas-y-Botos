# **Sucesiones**
Una suceión es una lista de numeros determinada por un patron que esta especificado en la expresión para el termino general $a_n$ de la sucesión. Por ejemplo,

$$b_{n}=2n$$
$$$$

$$c_{n}=1-\frac{1}{n}$$
$$$$

$$d_{n}=\frac{(-1)^n}{2^n}$$
$$$$

$$e_{n}=(-1)^n$$
$$$$

En los cuatro casos para $n=1,2,3,...$
$$$$

En general, nos interesa saber si la sucesion se aproxima o no a algun número particular.
$$$$



#Ejemplo 1

En la sucesión detrminada por $$b_{n}=2n$$
$$$$

podemos ver que cada termino de la sucesion es mayor que el anterior sin ningun valor límite,

```python
bn = [2*n for n in range(10)] # Aumente el numero de terminos en el rango para verlo
print (bn)
print(bn[5])                    # Mostrar el sexto término de la suceión
```

Esto puede verse claramente en la grafica de la sucesión

```python
plt.plot (bn,'.') # La sucesión crece sin límite
```

En este caso decimos que el límite no existe y escribimos

$$\lim_{n \to \infty}{b_n}=\infty$$
$$$$

#Ejemplo 2

En la sucesión determinada por $$c_{n}=1-\frac{1}{n}$$
$$$$

podemos ver que cada termino de la sucesion es mayor que el anterior, pero en este caso cada término se hace cada vez mas cercano a 1.

```python
cn = [ 1-(1/n) for n in range(1,50)] # Aumente el numero de terminos en el rango para verlo
print (cn)
```

Esto puede verse claramente en la grafica de la sucesión

```python
plt.plot(cn,'.') # La sucesion tiene un límite y es 1
```

En este caso decimos que el límite de la sucesion es 1 y escribimos

$$\lim_{n \to \infty}{c_n}=1$$
$$$$

$$\lim_{n \to \infty}{d_n}=0$$
$$$$

# **Propiedades de los límites de sucesiones**

Dadas $\{a_n\}$ y $\{b_n\}$ sucesiones con limites
$$$$

$$\lim_{n \to \infty}{a_n}=L$$
$$$$

$$\lim_{n \to \infty}{b_n}=M$$
$$$$

$b_n\neq0$, $M\neq0$, $k$ una constante y $p$ un número racional, entonces los siguientes enunciados son verdaderos,
$$$$

$i)$  $$\lim_{n \to \infty}k{a_n}=kL$$
$$$$

$ii)$  $$\lim_{n \to \infty}({a_n}+{b_n})=\lim_{n \to \infty}{a_n}+\lim_{n \to \infty}{b_n}=L+M$$
$$$$

$iii)$  $$\lim_{n \to \infty}({a_n}-{b_n})=\lim_{n \to \infty}{a_n}-\lim_{n \to \infty}{b_n}=L-M$$
$$$$

$iv)$  $$\lim_{n \to \infty}{a_n}{b_n}=(\lim_{n \to \infty}{a_n})(\lim_{n \to \infty}{b_n})=LM$$
$$$$

$v)$  $$\lim_{n \to \infty}\frac{a_n}{b_n}=\frac{\lim_{n \to \infty}{a_n}}{\lim_{n \to \infty}{b_n}}=\frac{L}{M}$$
$$$$

$vi)$  $$\lim_{n \to \infty}({a_n})^p=(\lim_{n \to \infty}{a_n})^p=L^p$$
$$$$


Verifiquemos estas propiedades utilizando las sucesiones que hemos tomado como ejemplo en este cuaderno.

$i)$  $$\lim_{n \to \infty}5{c_n}=\lim_{n \to \infty}5(1-\frac{1}{n})=5(1)=5$$
$$$$

Lo cual puede verse graficamente

```python
cn = [ 5*(1-(1/n)) for n in range(1,50)] # Aumente el numero de terminos en el rango para verlo
print (cn)
plt.plot(cn,'.')
```

$ii)$  $$\lim_{n \to \infty}({c_n}+{d_n})=\lim_{n \to \infty}{c_n}+\lim_{n \to \infty}{d_n}=1+0=1$$
$$$$

Lo cual puede verse en la gráfica

```python
Sn = [ (1-(1/n))+(((-1)**n)/(2**n)) for n in range(1,30)]  # suma
print (Sn)
plt.plot(Sn,'.')
```

$iii)$  $$\lim_{n \to \infty}({d_n}-{c_n})=\lim_{n \to \infty}{d_n}-\lim_{n \to \infty}{c_n}=0-1=-1$$
$$$$

De nuevo se verifica mediante la gráfica

```python
Rn = [ (((-1)**n)/(2**n))-(1-(1/n)) for n in range(1,30)]  # Resta
print (Rn)
plt.plot(Rn,'.')
```

$iv)$  $$\lim_{n \to \infty}{d_n}{c_n}=(\lim_{n \to \infty}{d_n})(\lim_{n \to \infty}{c_n})=0(1)=0$$
$$$$

Que se ve claramente al graficar

```python
Mn = [(((-1)**n)/(2**n))*(1-(1/n)) for n in range(1,30)]  # Producto
print (Mn)
plt.plot(Mn,'.')
```

$v)$  $$\lim_{n \to \infty}\frac{d_n}{c_n}=\frac{\lim_{n \to \infty}{d_n}}{\lim_{n \to \infty}{c_n}}=\frac{0}{1}=0$$
$$$$

La gráfica lo verifica

```python
Dn = [ (((-1)**n)/(2**n))/(1-(1/n)) for n in range(2,30)]  # Cociente
print (Dn)
plt.plot(Dn,'.')
```

$vi)$  $$\lim_{n \to \infty}({c_n})^{\frac{1}{3}}=(\lim_{n \to \infty}{c_n})^{\frac{1}{3}}=1^{\frac{1}{3}}=1$$
$$$$

y de nuevo se puede ver mediante la gráfica

```python
rn = [ (1-(1/n))**(1/3) for n in range(1,30)]  # Potencia
print (rn)
plt.plot(rn,'.')
```

# **Métodos analíticos para evaluar límites**

En algunos casos particulares, es posible evaluar el límite sin recurrir a tecnicas numéricas. Por ejemplo en el caso
$$$$

$$a_n=  \frac{n^2+7n+12}{3n^2+7}$$

Para evaluar el límite se divide numerador y denominador por la máxima potencia de $n$ y se obtiene,
$$$$

$$\lim_{n \to \infty}{a_n}=\frac{1}{3}$$
$$$$

Este procedimiento constituye un **método analítico** para determinar límites de sucesiones.
$$$$

# **Métodos numéricos para estimar límites**

En general, estimar límites de sucesiones es algo que requiere herramientas de computo, por ello se habla de métodos o técnicas numéricas de estimar límites. Estas técnicas requieren de una definición formal del límite de una sucesión y llevan al concepto de convergencia. A continuación se presenta una sencilla tecnica numérica para estimar el límite de la sucesión
$$$$

$$a_n=  \frac{n^2+7n+12}{3n^2+7}$$
$$$$

Y a continuación se discuten formalmente las ideas de límite y convergencia de una sucesión.

```python
# Tecnica numérica para estimar el límite de la sucesión (an)

fn = [ (n**2+7*n+12)/(3*n**2+7) for n in range(1,1000)]
print (fn)
plt.plot(fn,'.')     # El límite de esta sucesíon es 1/3
plt.ylim(0.3,0.4)
```

#**Definicion formal de límite**

Se dice que el límite de una sucesión $\{a_n\}$ es $L$, y se escribe
$$$$

$$\lim_{n \to \infty}{a_n}=L$$
$$$$

si para cada valor $\epsilon>0$ arbitrario, existe un entero $N$ tal que
$$$$

$$| a_n -L| < \epsilon$$  

para todo $n > N$.
$$$$


### **Forma alternativa**

Existe $N$ tal que para $n>N$ se cumple que

$$L-\epsilon < a_n < L+\epsilon$$
$$$$

#**Convergencia**

Si el límite de una sucesión $\{a_{n}\}$ existe y es $L$, la sucesión converge a $L$.

Si una sucesión $\{a_n\}$ crece indefinidamente,
$$$$

$$\lim_{n \to \infty}{a_n}=\infty$$
$$$$

la sucesión diverge.

# Ejemplo

Demuestre que la sucesión dada converge. Considere el caso particular $\epsilon = 0.001$.
$$$$

$$a_n=\frac{n^2+6n+5}{4n^2+1}$$
$$$$

# Solución

Se trata de hallar $N$ tal que, para $n>N$ se cumple que

$$L-\epsilon < a_n < L+\epsilon$$
$$$$

Lo primero es hallar $L$, si es que existe.

```python
an = [(n**2+6*n+5)/(4*n**2+1) for n in range(2000)]
print(an)
```

El limite existe y es,

$$L=\lim_{n \to \infty}{a_n}=0.25$$
$$$$

Ahora revisamos el intervalo de interés

$$L-\epsilon < a_n < L+\epsilon$$
$$$$

$$L-0.001< a_n < L+0.001$$
$$$$

$$0.25-0.001< a_n < 0.25+0.001$$
$$$$

$$0.249< a_n < 0.251$$
$$$$

```python
                        # Este procedimiento constituye un método geométrico para determinar convergencia de una sucesión
plt.plot(an,'.')
plt.ylim(0.249,0.251)
plt.axhline(0.25)
```

Ahora buscamos el indice $n$ más pequeño para el cual, el término $a_n$ y todos los siguientes términos  de la sucesión están contenidos en el intervalo
$$$$


$$0.249< a_n < 0.251$$
$$$$

mostrado en la figura generada en la celda de codigo anterior.
$$$$

```python
                     # Este procedimiento constituye un método numérico para determinar límites de sucesiones.
an = [(n**2+6*n+5)/(4*n**2+1) for n in range(5000)]                              # Se genera la sucesión mediante una lista
print (an)
```

```python
import numpy as np
e = np.array(an)-(0.25)                                                          # Se crea un arreglo con la lista (an) para poder hacer la
                                                                                 #  resta (an-L) y generar el arreglo e.
print (e)
```

```python
h=len(e)                                                                         #  Se indexa el arreglo e

for j in range(h):                                                               #  Se revisa en el rang (h) cual es el menor
  if e[j] < 0.001:                                                               #  indice que satisface la condición
    print(j)                                                                     #  y se muestra el índice
    break
```

A partir de $N=1501$ todos los términos de la sucesion $\{a_n\}$ estan en el intervalo,
$$$$

$$L-0.001< a_n < L+0.001$$
$$$$

$$0.249< a_n < 0.251$$
$$$$

Por consiguiente la sucesión,
$$$$

$$a_n=\frac{n^2+6n+5}{4n^2+1}$$
$$$$

converge a $L=0.25$.
$$$$

A continuación se presenta el codigo del proceso completo, basado en los pasos indicados hasta aqui.

```python
import matplotlib.pyplot as plt
import numpy as np

an = [(n**2+6*n+5)/(4*n**2+1) for n in range(25000)]
print(an)

plt.plot(an,'.')
plt.ylim(0.249,0.251)
plt.axhline(0.25)

an = [(n**2+6*n+5)/(4*n**2+1) for n in range(2000)]
print (an)

e = np.array(an)-(0.25)
print (e)

h=len(e)

for j in range(h):
  if e[j] < 0.001:
    print(j)
    break

```
